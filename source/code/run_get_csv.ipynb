```python
import csv
import os
import benford_torch as ref
from pathlib import Path

MANIFEST = Path("../scripts/manifest.csv")
LEVELS   = ["level1", "level2", "level3"]
N_IMAGES = 6500


def process_images():
    Q_80  = ref.get_quantization_matrix(80)
    Q_85  = ref.get_quantization_matrix(85)
    Q_90  = ref.get_quantization_matrix(90)
    Q_95  = ref.get_quantization_matrix(95)
    Q_100 = ref.get_quantization_matrix(100)

    quant_matrices = [Q_80, Q_85, Q_90, Q_95, Q_100]
    bases          = [10, 20, 40, 60]
    frequencies    = [1, 2, 3, 4, 5, 6, 7, 8, 9]

    divergences_str    = ["js", "r", "t"]
    bases_str          = ["10", "20", "40", "60"]
    frequencies_str    = ["1", "2", "3", "4", "5", "6", "7", "8", "9"]
    quant_matrices_str = ["80", "85", "90", "95", "100"]

    os.makedirs("csv", exist_ok=True)

    for split in ("train", "test"):
        os.makedirs(f"csv/{split}", exist_ok=True)

        ref.generate_csv_header(divergences_str, bases_str, frequencies_str, quant_matrices_str, f"csv/{split}/clean.csv", include_video=True)
        ref.process_and_save_images(f"../frames/{split}/clean", None, "0", bases, frequencies, quant_matrices, f"csv/{split}/clean.csv")

        ref.generate_csv_header(divergences_str, bases_str, frequencies_str, quant_matrices_str, f"csv/{split}/corrupted.csv", include_video=True)
        for level in LEVELS:
            ref.process_and_save_images(f"../frames/{split}/corrupted/{level}", None, "1", bases, frequencies, quant_matrices, f"csv/{split}/corrupted.csv")
```

```python
process_images()
```
